```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 1. Load Dataset
df = pd.read_csv("data.csv")
print("Jumlah Missing Values:\n", df.isnull().sum())

# 2. Preprocessing & Encoding Target
selected_features = [
    'mean radius', 'mean texture', 'mean perimeter', 'mean area',
    'mean smoothness', 'mean compactness', 'mean concavity', 'mean concave points'
]
X = df[selected_features]
y = df['diagnosis'].map({'B': 0, 'M': 1}) # 0: Jinak, 1: Ganas

# 3. Train-Test Split (70% Latih, 30% Uji)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
# 4. Pelatihan Model A (Tanpa Pruning / Potensi Overfit)
model_a = DecisionTreeClassifier(max_depth=None, random_state=42)
model_a.fit(X_train, y_train)

# 5. Pelatihan Model B (Dengan Pre-Pruning max_depth=3)
model_b = DecisionTreeClassifier(max_depth=3, random_state=42)
model_b.fit(X_train, y_train)

print(f"Model A - Akurasi Latih: {accuracy_score(y_train, model_a.predict(X_train)):.4f} | Akurasi Uji: {accuracy_score(y_test, model_a.predict(X_test)):.4f}")
print(f"Model B - Akurasi Latih: {accuracy_score(y_train, model_b.predict(X_train)):.4f} | Akurasi Uji: {accuracy_score(y_test, model_b.predict(X_test)):.4f}")

print("\n--- Classification Report Model B ---")
print(classification_report(y_test, model_b.predict(X_test), target_names=['Jinak (0)', 'Ganas (1)']))
```

```python
# 6. Visualisasi Pohon Keputusan Model B
plt.figure(figsize=(18, 9))
plot_tree(model_b, feature_names=selected_features, class_names=['Jinak', 'Ganas'], filled=True, rounded=True)
plt.title("Visualisasi Pohon Keputusan Model B (max_depth=3)")
plt.show()

# 7. Simpan Model Terbaik (Model B) dengan Joblib
joblib.dump(model_b, "model/decision_tree_model.pkl")
joblib.dump(selected_features, "model/feature_names.pkl")
print("Model berhasil disimpan!")
```
